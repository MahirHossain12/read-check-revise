# 01 · Study design

Can evidence presentation and reflection help beginner Python learners evaluate AI explanations, revise their answers, and apply concepts?

Run the setup cell first. See the repository README for ZIP and GitHub setup.

```python
from pathlib import Path
import subprocess
import sys
import tempfile
from zipfile import ZipFile

REPO_URL = ""
PROJECT_DIR = Path.cwd()

def find_project(directory):
    candidates = [directory, *sorted(path for path in directory.iterdir() if path.is_dir())]
    return next((path for path in candidates if (path / "src" / "pilot.py").is_file()), None)

project = find_project(PROJECT_DIR)
if project is None:
    if REPO_URL:
        project = Path(tempfile.mkdtemp(prefix="python-hci-")) / "repo"
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(project)], check=True)
    else:
        try:
            from google.colab import files
        except ImportError as exc:
            raise RuntimeError("Extract the repository and set PROJECT_DIR to its root.") from exc
        import io
        uploaded = files.upload()
        archives = [data for name, data in uploaded.items() if name.lower().endswith(".zip")]
        if len(archives) != 1:
            raise ValueError("Upload one project ZIP.")
        directory = Path(tempfile.mkdtemp(prefix="python-hci-"))
        with ZipFile(io.BytesIO(archives[0])) as archive:
            archive.extractall(directory)
        project = find_project(directory)
if project is None or not (project / "study" / "stimuli.json").is_file():
    raise FileNotFoundError("Project root not found. Use the ZIP supplied with these notebooks.")
sys.path.insert(0, str(project.resolve()))
from src import pilot
OUT = project / "outputs"
print("Project root:", project)
```

Every learner receives all three conditions across three different topics. A shows a fixed explanation. B adds documentation evidence. C adds both evidence and a required reflection. The primary outcome is correct rejection of the inaccurate claim. Compare B − A and C − B descriptively. The latter includes the effort and time needed for the reflection.

The six sequences balance topics and conditions across trial positions and topic–condition pairings. Assign two learners per sequence for a target of 12. Balance refers to the planned full allocation and may change after withdrawals.

```python
pilot.validate_design()
for sequence in range(1, 7):
    print(sequence, " | ".join(f"{topic} / {condition}" for topic, condition in pilot.schedule(sequence)))
allocation = pilot.allocation()
path = pilot.write_csv(OUT / "planned_allocation.csv", allocation)
print(f"{len(allocation)} planned slots → {path}")
```
